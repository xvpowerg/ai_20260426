```python
import requests #下載網頁
from bs4 import BeautifulSoup
url = r"https://zh.wikipedia.org/zh-tw/2026年电影"
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}
resq = requests.get(url,headers=headers)
resq.encoding="utf-8"
#print(resq.text)
soup = BeautifulSoup(resq.text,"html5lib")
movieTable = soup.find("table",class_="wikitable sortable")
tbody = movieTable.find("tbody")
trList = tbody.find_all("tr")
for inx in  range(1,len(trList)):
    print("排名:",trList[inx].find("th").text,end="")
    print("片名:",trList[inx].select_one("a").text)
    print("全球票房",trList[inx].find_all("td")[-1].text)
```

## 重設新陣列
+ resize 這個動作會依據原本的陣列再設定指定大小的新陣列。 
+ resize 動作的參數如下：
+ numpy.resize(arr1，shape1)
  + arr1：原本的陣列
  + shape1：新規劃的大小
+ 如果規劃的新陣列比較大，將會重新複製原有陣列的資料，填滿新的儲存格。
+ resize 這個動作***建立新的陣列***，而 reshape 則是依據原有的重新規劃，仍會受到原有陣列影響。
+ transpose()轉置

```python
import numpy as np
a = np.array([[1,2,3],
             [4,5,6]])
print(a.shape)
print(a)
b = a.reshape((1,6)) #共用資料
print(b)
print(b.shape)
b[0,0] = 999
print(b)
print("=======================")
print(a)
```

```python
import numpy as np
a = np.array([[1,2,3],
             [4,5,6]])
print(a.shape)
print(a)
print("="*20)
b = np.resize(a,(3,2))#不會複製
print(b)
print(b.shape)
b[0,0] = 990
print("="*20)
print(b)
print("="*20)
print(a)
```

```python
import numpy as np
a = np.array([[1,2,3],
             [4,5,6]])
print(a.shape)
print(a)
b = np.resize(a,(10,10))
print(b)
```

### transpose

```python
import numpy as np
newArray =  np.arange(10).reshape(2,5)
print("====================")
print(newArray)
print("====================")
print(newArray.transpose())
print("====================")
print(newArray.T)
print("====================")
tArray = newArray.T
tArray[1] = 9999
print(newArray)
```

## 總和最大與最小
+ np.sum( ) 代表某一個陣列內容的總和，也可以指定這個陣列的哪一軸 (axis) 內容總和。
+ np.min( ) 代表某一個陣列內容的最小值，也可以指定這個陣列的哪一軸 (axis) 內容的最小值。
+ np.max( ) 代表某一個陣列內容的最大值，也可以指定這個陣列的哪一軸 (axis) 內容的最大值。
+ amin 與 min 是相同功能的方法。
+ amax 與 max 是相同功能的方法。

```python
import numpy as np
a = np.array([[0,30,45],
              [60,75,90]])
print(a)
print("=======================")
b = np.sum(a)
print(b)
b = np.sum(a,axis=0)
print(b)
b =  np.sum(a,axis=1)
print(b)
```

### max() min()

```python
import numpy as np
a = np.array([[0,30,45],
              [60,75,90]])
print(a)
b = np.max(a)
print(b)
b = np.max(a,axis=0)
print(b)
b = np.max(a,axis=1)
print(b)
```

```python
import numpy as np
a = np.array([[0,30,45],
              [60,75,90]])
print(a)
bi = np.argmax(a)
print(bi)
bi = np.argmax(a,axis=0)
print(bi)
bi = np.argmax(a,axis=1)
print(bi)
```

```python
import numpy as np
a = np.array([[0,30,45],
              [60,75,90]])
print(a)
b = np.mean(a)
print(b)
b = np.mean(a,axis=0)
print(b)
b = np.mean(a,axis=1)
print(b)
```

```python
import numpy as np
a = np.array([1,2,3,4])
b = np.average(a)
print(b)
wts = np.array([4,3,2,1])
b = np.average(a,weights=wts)#(1*4+2*3+3*2+4*1)/(4+3+2+1)
print(b)
b = np.average([1,2,3,4],weights=[4,3,2,1],returned=True)
print(b)
```

```python
import numpy as np
a = np.array([[0,30,45],
              [60,75,90]])
print(a)
b = np.median(a)#(45+60)/2
print(b)
b = np.median(a,axis = 0)
print(b)
b = np.median(a,axis =1)
print(b)
```
