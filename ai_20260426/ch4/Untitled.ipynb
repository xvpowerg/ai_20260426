## 自訂函式的運用

#### 程式設計時儘量減少相同語法的重複出現，為什麼?

修改時可能忘記其他地方沒有改，導致程式執行出現錯誤結果。

減少重複語法對於之後的維修會比較方便。


#### 要如何減少重複的相同語法?

把重複的相同語法當作工具看待，需要的時候呼叫他。

這種工具於程式語法內叫做函數(function)。


```python
def func_test():
    print("呼叫函式")
func_test()
func_test()
func_test()    
```

## 變數影響範圍
+ 函數外的變數：
  + 函數內可以顯示該變數內容
  + 不屬於函數的區域內都可以使用
+ 函數內的變數：
  + 只在函數內產生效果，不會影響函數外的變數
  + 若函數內沒有進行變數宣告而進行改變內容動作將會產生錯誤訊息!

```python
a = 5
def func_test1():
    a = 10
    print("函式內:",a)
print("函式外1:",a)    
func_test1()
print("函式外2:",a)    
```

```python
i = 10
def checkFunc():
    print("內部:",i)
print("外部:",i)    
checkFunc()
```

```python
i = 10
def checkFunc():
    i = i + 1
    print("內部:",i)
print("函數外 1:",i)    
checkFunc()
print("函數外 2:",i)
```

```python
i = 10
def checkFunc():
    global i #宣告使用外部的i
    i = i + 1
    print("內部:",i)
print("函數外 1:",i)    
checkFunc()
print("函數外 2:",i)
```

### 不固定接收數量
( ) 內接收資料若是以 * 表示代表可以引入不定數量的參數：

一個*代表以 tuple 的方式引入

兩個*代表以 dict 的方式引入

```python
def printInfo(arg1,arg2="default",*arg3):
    print("arg1:",arg1)
    print("arg2:",arg2)
    print("arg3:",arg3)
    print("===================================")
printInfo(1)    
printInfo(1,2)
printInfo(1,2,3)
printInfo(1,2,3,4) 
printInfo(1,2,3,4,5,6,7) 

```

```python
def printInfo(*arg1,arg2,arg3="default"):
    print("arg1:",arg1)
    print("arg2:",arg2)
    print("arg3:",arg3)
printInfo(1,5,6,7,arg2 = "Ken",arg3="Hello!")    
```

```python
def sumValue(*arg):
    total = 0
    for v in arg:
        total += v
    return total    
    
print(sumValue())#0
print(sumValue(1,2,3))#6
print(sumValue(1,2,3,4,5))#15
```

```python
#請使用cType表示計算方式
def caluculate(a,b,**args):    
    if "cType" not in args:
        print("請填入cType")
        return
    cType = args["cType"]
    if cType == "+":
        return a + b
    elif cType == "-":
        return a - b
    elif cType == "*":
        return a * b
    elif cType == "/":
        return a / b
print(caluculate(6,2,cType="/"))
    
```

## 匿名函數
不需要定義函數名稱，只需要用到運算式或表達分析語法。

Python使用lambda語法定義匿名函數。

匿名函數是一個表達式或計算式，並不是一個執行流程區塊。

匿名函數出現在一般函數不允許的地方，例如list裡面或著函數呼叫參數的位置。


```python
def f(x,y,z):
    return x+y+z
print(f(2,30,50))
print(type(f))
f2 = lambda x,y,z:x + y + z
print(f(2,30,50))
print(type(f2))
```

```python
list2 = [lambda x:x **2,lambda x:x**3,lambda x :x ** 4]
for func in list2:
    print(func(3))
```
