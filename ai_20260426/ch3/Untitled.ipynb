## 作業 文字接龍

```python
txt = input("請輸入第一個字串")
strList = []
strList.append(txt)
msg = txt
count = 0
inputMsg = "請輸入-"+txt[-1]+"-開始的字串"

while count < 5:
    tmpTxt = input(inputMsg)
    if tmpTxt[0] == txt[-1]:
        txt = tmpTxt
        strList.append(tmpTxt)
        msg = "-".join(strList)
        inputMsg = "請輸入-"+txt[-1]+"開始的字串"
    else:
        inputMsg = "請再次輸入-"+txt[-1]+"開始的字串"
        count += 1
    print("上一個字串是:"+msg)    
    
```

## list 項目擴展資料
+ extend 擴展資料：
  + 如果引號括起來的字串有 [ ] 將會拆解成好幾個資料。
  + 可以把其他 list 加入到這個 list 內，擴展多個欄位。

```python
list1 = [20,30,40,50]
list2 = [55,66,77]
print(list1)
list1.extend(list2)
print(list1)
```

```python
list1 = [20,30,40,50]
list2 = [55,66,77]
print(list1)
list1 += list2
print(list1)
```

## list 項目插入新資料
+ list 可使用 insert 方式插入新的資料：
   + list.insert(索引值，插入的資料)。
   + 若索引值超出範圍則會改為 append 方式附加。

```python
list1 = ["a","b","c","d","e"]
print(list1)
list1.insert(1,"x")
print(list1)
list1.insert(-2,"y")
print(list1)
list1.insert(88,"g")
print(list1)
list1.insert(-99,"yy")
print(list1)
list1.insert(-1,"zz")
print(list1)
```

## 移除 
+ list 可透過以下三種方式移除項目：
  + list.remove( )
  + list.pop( )
  + del list[ ]

```python
list1 = ["1","x","2","x","3"]
list1.remove("x")
print(list1)
list1.remove("x")
print(list1)
key = "x"
if key in list1:#list1內有如果有x
    list1.remove("x")#沒有x了所以出錯
```

```python
list1 = ["1","x","2","x","3"]
#使用while 迴圈 幫我一次性移除所有x
removeKey = "x"
while removeKey in list1:
    list1.remove(removeKey)
print(list1)
```

### pop

如果 pop( ) 內沒有參數，則移除最後一筆資料。

pop() 內若有參數，則移除指定位置的資料。

```python
list2 = []
list2.append("Ken")
list2.append("Vivin")
list2.append("Lindy")
list2.append("Joy")
print(list2)
v1 = list2.pop()#最後一筆 FILO 先進後出
print("v1:",v1)
print(list2)
v2 = list2.pop()#最後一筆 FILO 先進後出
print("v2:",v2)
print(list2)
v3 = list2.pop()#最後一筆 FILO 先進後出
print("v3:",v3)
print(list2)
v4 = list2.pop()#最後一筆 FILO 先進後出
print("v4:",v4)
print(list2)
v5 = list2.pop()#會出錯 因為空了不可pop
```

```python
list2 = []
list2.append("Ken")
list2.append("Vivin")
list2.append("Lindy")
list2.append("Joy")
print(list2)

while len(list2) > 0:
    print(list2.pop())
```

```python
list3 = ["A","B","C","D"]
v1 = list3.pop(0)
print(list3)
print("v1:",v1)
v2 = list3.pop(0)
print(list3)
print("v2:",v2)
```

### Del
+ list 可透過 del 方式移除指定位置資料。
+ del list[ ] 必須指定範圍：
  + del list[n] 代表移除第 n 位索引值資料。
  + del list[:m] 代表移除第 m 位索引值之前 ( 不包含 m ) 資料。
  + del list[n:m] 代表移除第 n 到 m 位索引值之前 ( 不包含 m ) 資料。

```python
list3 = ["A","B","C","D","E","F","G"]
print(list3)
del list3[1]
print(list3)
del list3[1:4]
print(list3)
```

## Dictionary

資料結構是由 key:value 所組成。

key 不能夠重複，否則會被後面的結果蓋過去。

可輸入 key 找尋您要找出來的值。

如果輸入的 key 不存在，那就會出現錯誤訊息。

key 只能是 字串 數字 tuple

```python
dict1 = {"a":100,"b":200,"c":300}
print(dict1)
print(dict1["b"])
print(dict1["a"])
print(dict1["x"])#KeyError x不存在出錯
```

```python
dict1 = {"a":100,"b":200,"c":300}
for k in dict1:
    print(k,":",dict1[k])
```
