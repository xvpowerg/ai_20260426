## 比較運算子

```python
a = 10
b = 5
ans = a > b
print(a,">",b,"=",ans)
ans = a < b
print(a,"<",b,"=",ans)
ans = a >= b
print(a,">=",b,"=",ans)
ans = a <= b
print(a,"<=",b,"=",ans)
ans = a == b#是否相等
print(a,"==",b,"=",ans)
ans = a!=b#不等於
print(a,"!=",b,"=",ans)
```
