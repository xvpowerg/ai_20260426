設計錢包總金額為 1000 元，

如果購買商品總金額超過 1000 

會引發一個金額不足的例外 OverflowError 

```python
mySum = 0
dict1 = {}
try:
    while True:
        item = input("消費項目")
        price = input("消費金額")
        if mySum + int(price) > 1000:
            raise OverflowError("已經超支了")
        dict1[item] = price
        mySum += int(price)
finally:
    print(dict1)    
    
```

## Json

## 關於檔案關閉動作
+ 檔案開啟後須進行 close( ) 方法進行關閉動作。
+ 若檔案沒關閉會造成：
  + 開啟的文件物件會占用系統資源。
  + Python 可以同時間開啟的文件數量有限制 (約 20 份文件)。
  + 開啟文件物件的模式若為寫入 (w 或 a) 模式，一般都是暫存於緩衝區，系統閒置或文件關閉前才會進行寫入，若沒有進行 close( ) 動作可能造成文件儲存不完整。

```python
import os
class TestFile:
    def openFile(self,filePath):
        self.handle = open(filePath,"w")
        self.handle.close()
t = TestFile()
t.openFile("test.txt")
os.remove("test.txt")
print("Success")
```

```python
import os
class TestFile:
    def openFile(self,filePath):
        with open(filePath,"w") as f:
            f.write("Hello!")
t = TestFile()
t.openFile("test.txt")
os.remove("test.txt")
print("Success")
```

## BeautifulSoup Base
find() 只找第一個符合條件的

https://www.crummy.com/software/BeautifulSoup/bs4/doc.zh/#find-all

find_all() 找所有符合條件的

select_one() 只找一個符合條件的可直接使用css語法

select() 找所有符合條件的可使用css語法


```python
html='''<a id='a1' class="c1">V1</a>
<a id='a2' class="c1">V2</a>
<a id='a3' class="c3">V3</a>
'''
soup = BeautifulSoup(html,"html5lib")
boj1 = soup.find(id="a1")
print(boj1)
boj2 = soup.find_all(class_="c1")
print(boj2)
obj3 = soup.select_one("#a2")
print(obj3)
obj4 = soup.select(".c1")
print(obj4)
```

```python
from bs4 import BeautifulSoup
html='''<a id='a1' class="c1 c2" href="https://google.com">Google</a>
<a id='a2' class="c1"  href="https://yahoo.com">Yahoo</a>
<a id='a3' class="c3" href="https://facebook.com">Facebook</a>
'''
soup = BeautifulSoup(html,"html5lib")
txt = soup.find(id="a1").getText()
print(txt)
txt2 = soup.find(id="a2").text
print(txt2)
href = soup.find(id="a1").get("href")
print(href)
classList = soup.find(id="a1").get("class")
print(classList)

tmpList = soup.find_all(class_="c1")
for tag in tmpList:
    print(tag.get("href"))
```

```python
import bs4
htmlFile = open("web1.html",encoding="utf-8")
objSoup = bs4.BeautifulSoup(htmlFile,"html5lib")
objTag = objSoup.find_all("div")
"""
.string 會返回標籤最裡面的內容，如果標籤內還有其他子節點(若有子標籤，代表不是單純的字串)，而會輸出 None。
"""
for data in objTag:
    print("data:",data)
    print("data string:",data.string)
```

```python
import bs4
htmlFile = open("web1.html",encoding="utf-8")
objSoup = bs4.BeautifulSoup(htmlFile,"html5lib")

"""
.string 會返回標籤最裡面的內容，如果標籤內還有其他子節點(若有子標籤，代表不是單純的字串)，而會輸出 None。
repr(）方法可將物件或其他資料轉換為字串進行輸出!
"""

for data in objSoup.strings:
    print("data:" ,repr(data))
```
