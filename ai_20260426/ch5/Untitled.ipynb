```python
priceList = []
priceSum = 0
for i in range(1,5):
    p = input(f"請輸入{i}個月的支出金額")
    p = int(p)
    priceSum += p
    priceList.append(p)
priceList.sort()
print("Max:",priceList[-1])
print("Min:",priceList[0])
print("priceSum:",priceSum)
print(priceList)
```

## 例外語法
try:
    保護的程式區段
    
except 例外1 :
    例外1發生時執行的程式
    
except :
    其他例外發生時執行的程式
    #未指定例外型態，捕捉所有例外物件
    
else:
    若try程式區段沒產生例外執行的程式
    
finally:
    不管有沒有發生例外都會執行的程式

```python
num1 = 10
num2 = 0
print("開始")
try:
    print(num1 / num2)
except:
    print("有錯誤")
print("完成")
```

```python
num1 = 10
num2 = 1
myList = [1,2,3,4,5]
print("開始")

try:
    print(num1 / num2)#ZeroDivisionError
    print(myList[9])#IndexError
except ZeroDivisionError:    
    print("分母不可為零")
except IndexError:
    print("Index錯誤")
print("完成")
```

```python
num1 = 10
num2 = 0
myList = [1,2,3,4,5]
print("開始")

try:
    print(num1 / num2)#ZeroDivisionError
    print(myList[0])#IndexError
    #aaaa
except ZeroDivisionError as ex:    
    print("分母不可為零:",ex)
    
except IndexError as ex:
    print("Index錯誤:",ex)
except Exception as ex:
    print("有錯誤:",ex)
print("完成")
```

## 使用raise拋例外

```python
score = 950
def testScore(score):
    if score > 100 or score < 0:
        raise OverflowError("錯誤的成績")
    print(score)    
print("開始")    
testScore(score) 
print("完成")    
```

```python
class Person:
    #共享
    company = ""
    def __init__(self,firstName,lastName):
        self.firstName = firstName
        self.lastName = lastName
    def printInfo(self):    
        print("Company:",Person.getCompany()," FirstName:",self.firstName," LastName:",self.lastName)
    @classmethod    
    def setCompany(cls,company):
        cls.company = company
    @classmethod    
    def getCompany(cls):    
        return cls.company
Person.setCompany("IBM") 
p1 = Person("Ken","Le")        
p1.printInfo()
p2 = Person("Tom","AA")
p2.printInfo()
       
```

## 私有化

```python
class Animal:
    def __init__(self,name,age):
        self.name = name
        #私有化
        self.__age = age
    def printInfo(self):        
        print(self.name,self.__age)
    def setAge(self,age):
        if age < 0 or age > 200:
            raise OverflowError("錯誤的年齡")
        self.__age = age
    def getAge(self):    
        return self.__age
a1 = Animal("Bobo",3)        
a1.printInfo()
a1.setAge(-25)
a1.printInfo()
#作業name私有化
#name長度不可小於2 不可大於20
```
